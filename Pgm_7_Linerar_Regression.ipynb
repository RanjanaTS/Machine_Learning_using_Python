```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

x = np.array([1, 2, 3, 4, 5]).reshape(-1, 1) # Reshape x for scikit-learn
y = np.array([40, 50, 60, 70, 80])

model = LinearRegression()
model.fit(x, y)
pred_y = model.predict(x)
print(pred_y)

plt.scatter(x, y, label="Actual data")
plt.plot(x, pred_y, label="Predicted data")
plt.xlabel("Marks Scored")
plt.ylabel("Score") # Added a y-label for clarity
plt.title("Linear Regression Prediction") # Added a title
plt.legend()
plt.show() # Display the plot
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([20,35,50,65,80])

model=LinearRegression()
model.fit(x,y)
pred_m=model.predict([[6]])
print("predicted marks: ",pred_m)
print("slope: ",model.coef_)
print("intercept: ",model.intercept_)
```
