```python
# second largest element in a list
l=[1,2,3,4,10,70]
print(max(l))
s=sorted(l)
print(s[-2])
```

```python
# remove duplicates in a list
l=[1,2,3,4,4,4,5,1]
print(list(set(l)))
```

```python
# common elements in two list
a=[1,2,3,4,5]
b=[4,5,6,7,8]
common=[]
for i in a:
    for j in b:
        if i==j:
            common.append(i)
print(common)
```
