```python
# CREATING A DICTIONARY
d1 = {1:"apple",2:"orange"}
print(d1)
d2={'name':'ranjana','age':19,'qualification':'AIML'}
print(d2)
```

```python
# ACCESSING A VALUE IN DICTIONARY
d2={'name':'ranjana','age':19,'qualification':'AIML'}
print(d2["name"])
print(d2["age"])
```

```python
# ADDING A NEW KEY:VALUE PAIR
d2={'name':'ranjana','age':19,'qualification':'AIML'}
d2['city']='coimbatore'
print(d2)
```

```python
# UPDATING AN EXISTING VALUE
d2={'name':'ranjana','age':19,'qualification':'AIML'}
d2['age']=20
print(d2)
```

```python
#DELETING A KEY:VALUE PAIR
d2={'name':'Sureka','age':19,'qualification':'AIML'}
del d2['age']
print(d2)
```

```python
d2={'name':'ranjana','age':19,'qualification':'AIML'}
d2.pop('qualification')
print(d2)
```

```python
#DISPLAYING ALL KEYS,VALUES AND KEY:VALUE PAIRS
d2={'name':'ranjana','age':19,'qualification':'AIML'}
print(d2.keys())
print(d2.values())
print(d2.items())
```

```python
# TRAVERSING THE DICTIONARY
d3={'name':'ranjana','age':19,'gender':'female','qualification':'AIML'}
for i,j in d3.items():
    print(i,":",j)
```
