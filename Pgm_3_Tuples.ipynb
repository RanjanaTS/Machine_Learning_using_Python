```python
# reverse a tuple
a=(1,2,3,4,5)
print(a[::-1])
```

```python
# removing duplicates 
a=(1,1,2,3,4,5,4,4)
print(tuple(set(a)))
```

```python
# appending element in list inside a tuple
b=(3,4,[1,2],5)
print(b[2])
b[2].append(3)
print(b)
```

```python
# factoial using math module
import math as m
print(m.factorial(5))
```

```python
# factorial
n=5
def fact(n):
    if n==0 or n==1:
        return 1
    return n*fact(n-1)
print(fact(n))
```

```python
# merge two tuples
t1=(1,2,3)
t2=(4,5,6)
t3=t1+t2

print(t3)
```

```python
#Tuples are immutable. 
# But using slicing and concatenation, we can create a new tuple with the desired changes and reassign it to the same variable
a=(100,222,300,400,555)
print("original tuple : ",a)
a=a[:1]+(200,)+a[2:4]+(500,)
print("modified tuple : ",a)
a=a[0:]+(600,700,)
print("modified tuple : ",a)
```
